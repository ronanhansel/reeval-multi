```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from factor_analyzer import FactorAnalyzer

df = pd.read_pickle('../data/resmat.pkl')
# NOTE: For your ordinal survey data, you would compute a polychoric correlation matrix first.
# This step is computationally intensive and requires a specialized library like `pingouin`.
# The rest of this example uses a standard Pearson correlation, but for your real analysis,
# you would pass the polychoric matrix to the FactorAnalyzer instead of the raw dataframe.

# --- Step 2: Perform Factor Analysis to Get Eigenvalues ---
# We create a FactorAnalyzer object. We don't need to specify factors yet.
fa = FactorAnalyzer(rotation=None, n_factors=df.shape[1])
fa.fit(df)

# Get the eigenvalues
ev, v = fa.get_eigenvalues()

# --- Step 3: Apply the Criteria & Create the Scree Plot ---
# **Kaiser Criterion**
kaiser_criterion_factors = sum(1 for i in ev if i > 1)
print(f"Kaiser Criterion suggests retaining: {kaiser_criterion_factors} factors")

# **Scree Plot**
plt.figure(figsize=(10, 6))
# The x-axis is the number of factors (from 1 to number of variables)
plt.plot(range(1, df.shape[1] + 1), ev, 'o-', color='royalblue')
plt.title('Scree Plot', fontsize=16)
plt.xlabel('Factor Number', fontsize=12)
plt.ylabel('Eigenvalue', fontsize=12)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# Add the Kaiser criterion line
plt.axhline(y=1, color='red', linestyle='--', label='Kaiser Criterion (Eigenvalue=1)')
plt.legend()
plt.show()
```
