```python
mapping = pd.read_csv('../data-reeval-multi/CORE_hard/model_mapping.csv')
# Filter mapping to only rows where 'Available Model' is not null or empty
filtered_mapping = mapping[mapping['Available Model'].notnull() & (mapping['Available Model'] != '')]
# Print the CORE Bench Model names that have a mapping
print(len(filtered_mapping['CORE Bench Model'].tolist()))
```

```python
corebench_hard_leaderboard = pd.read_csv('../data-reeval-multi/CORE_hard/corebench_hard_leaderboard.csv')
filtered_models = filtered_mapping['CORE Bench Model'].tolist()
filtered_leaderboard = corebench_hard_leaderboard[corebench_hard_leaderboard['Primary Model'].isin(filtered_models)]
filtered_leaderboard
```

```python
import re

def extract_percentage(acc):
    match = re.match(r'(\d+(\.\d+)?)%', acc)
    return float(match.group(1)) if match else None

leaderboard = filtered_leaderboard[['Primary Model', 'Accuracy']][filtered_leaderboard['Scaffold'] == 'CORE-Agent'].copy().rename(columns={'Primary Model': 'model', 'Accuracy': 'accuracy'})

leaderboard['accuracy'] = leaderboard['accuracy'].apply(extract_percentage)
leaderboard.to_csv('../data-reeval-multi/CORE_hard/corebench_hard_leaderboard_filtered.csv', index=False)
```
