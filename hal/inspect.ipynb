```python
import pandas as pd
```

```python
df = pd.read_pickle('data/all_benchmarks_inputs.pkl')
df
```

```python
# Requires transformers>=4.51.0
# Requires sentence-transformers>=2.7.0

from sentence_transformers import SentenceTransformer

# Load the model
model = SentenceTransformer("Qwen/Qwen3-Embedding-4B")

# Get task inputs from the dataframe
task_inputs = df['task_input'].tolist()

# Encode all task inputs to embeddings
# The model will output embeddings (dimension depends on model, typically 1024 or higher)
embeddings = model.encode(task_inputs, show_progress_bar=True)

print(f"Embedded {len(embeddings)} task inputs")
print(f"Embedding dimension: {embeddings.shape[1]}")
print(f"Embeddings shape: {embeddings.shape}")

# Store embeddings back in the dataframe
df['embeddings'] = list(embeddings)
df.to_pickle('data/all_benchmarks_with_embeddings.pkl')
```
