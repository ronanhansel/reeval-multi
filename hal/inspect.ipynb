```python
# Requires transformers>=4.51.0
# Requires sentence-transformers>=2.7.0

import os
from sentence_transformers import SentenceTransformer
from tqdm import tqdm

# Set cache directory to a local folder with write permissions
cache_dir = '/home/azureuser/cloudfiles/code/reeval-multi/hal/.cache/huggingface'
os.makedirs(cache_dir, exist_ok=True)
os.environ['TRANSFORMERS_CACHE'] = cache_dir
os.environ['HF_HOME'] = cache_dir

# Load the model with explicit cache directory
model = SentenceTransformer("Qwen/Qwen3-Embedding-4B", cache_folder=cache_dir)

# Get task inputs from the dataframe
task_inputs = df['task_input'].tolist()

# Encode all task inputs to embeddings
# The model will output embeddings (dimension depends on model, typically 1024 or higher)
embeddings = model.encode(task_inputs, show_progress_bar=True)

print(f"Embedded {len(embeddings)} task inputs")
print(f"Embedding dimension: {embeddings.shape[1]}")
print(f"Embeddings shape: {embeddings.shape}")

# Store embeddings back in the dataframe with progress bar
tqdm.pandas(desc="Storing embeddings")
df['embeddings'] = list(embeddings)
df.to_pickle('data/all_benchmarks_with_embeddings.pkl')
```
